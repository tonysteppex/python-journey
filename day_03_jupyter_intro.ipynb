```python
minutes_studied = [30, 45, 60, 90, 120]

total_minutes = sum(minutes_studied)
average_minutes = total_minutes / len(minutes_studied)

print(f"Total study time: {total_minutes} minutes")
print(f"Average session: {average_minutes} minutes")
print(f"Total hours: {total_minutes / 60:.2f}")
```
