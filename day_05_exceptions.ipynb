```python
try:
    x = int(input("What is the value of x? "))
    print(f"The value of x is: {x}")
except ValueError:
    print("Invalid input. Please enter an integer.")
```

```python
try:
    x = int(input("What is the value of x? "))
except ValueError:
    print("Invalid input. Please enter an integer.")
else:
    print(f"The value of x is: {x}")
```

```python
while True:
    try:
        x = int(input("What is the value of x? "))
    except ValueError:
        print("Invalid input. Please enter an integer.")
    else:
        break

print(f"The value of x is: {x}")
```

```python
def main():
    x = get_int("What's x? ")
    print(f"x is {x}")

def get_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            pass

main()
```
